```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
pi = 3.1415
t = np.linspace(-5, 15, 1000)
```

```python
taui = 1
def s(t):
    return 1*(abs(t)<taui/2)
plt.plot(t,s(t))
```

```python
def integrand(t,tau):
    return s(t)*s(t-tau)

tau = np.linspace(-2.5,2.5,1000)
VKFpr = np.linspace(-10,10,1000)
def VKF(tau):
    return (quad(integrand, -np.inf, np.inf, args=(tau))[0])        #Ошибка - это фейк, не верь, тут всё как нужно!
                                                                    
for i in range (len(tau)):
    VKFpr[i] = VKF(tau[i])
    
def cm_to_inch(value):
    return value/2.54

plt.figure(figsize=(cm_to_inch(40),cm_to_inch(10)))
plt.plot(tau,VKFpr)
```
