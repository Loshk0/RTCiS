Задание 1 Разложение прямоугольного импульса в ряд Котельникова

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
pi = 3.1415
t = np.linspace(-2, 2, 500)
```

```python
taui = 1
def s(t):
    return 1*(abs(t)<taui/2)
plt.plot(t,s(t))
```

```python
n=15   #n - количество лепестков (определяет верхнюю частоту)
wv = 2*pi*n/taui
Ncount = 30 #количество отсчётов
count = list(range(-Ncount,Ncount,1))
def snew(t):
    sum=0
    for i in count:
        sum = sum+s(i*pi/wv)*np.sin(wv*(t-i*pi/wv))/(wv*(t-i*pi/wv))
    return sum
plt.plot(t,snew(t))
plt.plot(t,s(t))
```
