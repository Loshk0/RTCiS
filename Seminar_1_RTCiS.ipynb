Задание 1 Синусоида

```python
import numpy as np
import matplotlib.pyplot as plt
pi = 3.142
t = np.linspace(-5e-3, 5e-3, 1000)
```

```python
A0 = 1
f0 = 1e3
w0 = 2*pi*f0
fi0 = pi
def s0(x):
    return A0*np.sin(w0*t+fi0) 
```

```python
plt.plot(t,s0(t))
```

Задание 2 Единичный видеоимпульс

```python
def sv(t):
    
    return 1 * (abs(t) < 0.002)
```

```python
plt.plot(t,sv(t))
```

Задание 3 Радиоимпульс

```python
sr= sv(t)*s0(t)
plt.plot(t,sr)
```
